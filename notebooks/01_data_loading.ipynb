## Column meanings

**matches.csv** (one row per match)
- `match_id` — unique match id
- `season` — tournament season (mixed formats, e.g. `2024` or `2007/08`)
- `city`, `venue` — location of the match
- `date` — match date
- `team1`, `team2` — the two competing teams
- `toss_winner`, `toss_decision` — toss outcome (bat/field)
- `winner` — match winner (blank if no result, e.g. washed out)
- `result_type` — which result fields were populated in the source data (not a clean category — reworked during cleaning)
- `player_of_match` — man of the match
- `event_name`, `match_type`, `gender` — tournament/match metadata
- `overs`, `balls_per_over` — format info (20, 6 for standard T20)
- `team1_won`, `toss_winner_won` — precomputed booleans in the source data (recomputed ourselves during preprocessing instead of trusted as-is)

**deliveries.csv** (one row per ball bowled)
- `match_id` — foreign key to matches.match_id
- `innings` — 1st or 2nd innings
- `batting_team`, `bowling_team`
- `over`, `ball_in_over`, `ball_number` — over and ball position
- `batter`, `bowler`, `non_striker`
- `batter_runs`, `extra_runs`, `total_runs`
- `wides`, `noballs`, `byes`, `legbyes`, `penalty` — extras breakdown
- `is_wicket`, `wicket_player_out`, `wicket_kind` — dismissal info
- `is_powerplay`, `is_middle_overs`, `is_death_overs` — phase-of-innings flags
- `match_winner`, `toss_winner`, `toss_decision`, `team1`, `team2` — match-level context duplicated onto every ball
- `batting_team_won` — whether the batting team on this ball went on to win
